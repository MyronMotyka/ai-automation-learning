```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
print(soup.title)
```

```python
print(soup.title.text)
```

```python
links = soup.find_all("a")
```

```python
print(links)
```

```python
print(len(links))
```

```python
for link in links:
    print(link.get_text(strip=True))
```

```python
for link in links:
    print("TEXT:", link.get_text(strip=True))
    print("URL:", link.get("href"))
    print("--------------------")
```
