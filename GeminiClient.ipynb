```python
client_name = input("Enter client name: ")
service = input("Enter service: ")
hours = float(input("Enter number of hours: "))
materials = float(input("Enter materials cost before tax: "))

hourly_rate = 65

labour = hours * hourly_rate
subtotal = labour + materials
hst = subtotal * 0.13
total = subtotal + hst

print(f"Client: {client_name}")
print(f"Service: {service}")
print(f"Labour: ${labour:.2f}")
print(f"Materials: ${materials:.2f}")
print(f"Subtotal: ${subtotal:.2f}")
print(f"HST: ${hst:.2f}")
print(f"Total: ${total:.2f}")
```

```python
job = {
    "client": "Anna",
    "service": "Furniture assembly",
    "hours": 2,
    "materials": 40,
    "total": 192.10,
    "city": "Ottawa",
}

print(job)

print("Client:", job["client"])
print("Service:", job["service"])
print("Hours:", job["hours"])
print("Total:", job["total"])
print("City:", job["city"])
```

```python
import json

job_json = json.dumps(job, indent=4)

print(job_json)
```

```python
restored_job = json.loads(job_json)

print(restored_job)
print("Client:", restored_job["client"])
print("Service:", restored_job["service"])

print("Type of job_json:", type(job_json))
print("Type of restored_job:", type(restored_job))
```

```python
prompt = f"""
You are an assistant for Myrofix Ottawa.

Here is the job information:
{job_json}

Write a short and polite estimate message for the client in English.
Include the service and total price.
"""

print(prompt)
```

```python
from google import genai
from getpass import getpass

api_key = getpass("Paste your Gemini API key: ")

client = genai.Client(api_key=api_key)

print("Gemini is connected")
```

```python
response = client.models.generate_content(
    model="gemini-3.7-flash",
    contents=prompt
)

print(response.text)
```
