```python
lead = {
    "name": "John Smith",
    "email": "john@example.com",
    "service": "Electical",
    "description": "I need two IKEA cabinets assembled."
}

print(lead)
```

```python
print(lead["name"])
print(lead["service"])
print(lead["description"])
```

```python
if lead["service"] == "Furniture Assembly":
    print("ACCEPT")
else:
    print("REVIEW")
```

```python
allowed_services = [
    "Furniture Assembly",
    "Door Installation",
    "Drywall Repair"
]

print(allowed_services)
```

```python
if lead["service"] in allowed_services:
    print("ACCEPT")
else:
    print("REVIEW")
```

```python
leads = [
    {
        "name": "John",
        "service": "Furniture Assembly"
    },
    {
        "name": "Anna",
        "service": "Electrical"
    },
    {
        "name": "Mike",
        "service": "Door Installation"
    }
]
```

```python
print(leads)
```

```python
print(leads[0])
```

```python
print(leads[1]["name"])
```

```python
for lead in leads:
    print(lead["name"])
```

```python
for lead in leads:

    if lead["service"] in allowed_services:
        print(lead["name"], "-", "ACCEPT")
    else:
        print(lead["name"], "-", "REVIEW")
```

```python
url = "https://jsonplaceholder.typicode.com/users"
```
